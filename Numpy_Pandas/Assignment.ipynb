```python
data = {
    "Name": ["A", "B", "C", "D", "E", "F", "G", "H", "I", "J"]
}
df = pd.DataFrame(data)
```

```python
print(df[df["CGPA"] == df["CGPA"].max()])
print(df["CGPA"].min())
print(df["Name"])
```
