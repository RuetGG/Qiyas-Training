```python
import pandas as pd
```

```python
data = {
    "Name": ["Abel", "Hana", "Robel"],
    "Math": [80, 90, 70]
}
df = pd.DataFrame(data)
print(df["Math"].mean())
print(df["Math"].max())
print(df["Math"].min())
```

```python
data = {
    "Name":  ["Abel", "Hana", "Robel"],
    "Salary": [1000, 2000, 3000]
}
df = pd.DataFrame(data)
print(df["Salary"].mean())
print(df["Salary"].max())
print(df["Salary"].min())
```

```python
# print(df["Salary"] >= 2000)
print(df[df["Salary"] >= 2000])
```

```python
data = {
    "Course": ["CS", "IT", "SE", "DS"],
    "Students": [120, 100, 90, 80]
}
df = pd.DataFrame(data)
print(df["Students"].sum())
print(df.loc[df["Students"].idxmax()])
```
