```python
import pandas as pd
```

```python
data = {
    "Name": ["Abel", "Hana", "Robel", "Ruth"],
    "Math": [80, 95, 60, 85],
    "Physics": [70, 88, 75, 90],
    "Chemistry": [90, 91, 70, 95]
}
df = pd.DataFrame(data)

df["Total"] = (
    df["Math"] + df["Physics"] + df["Chemistry"]
)

df["Average"] = df["Total"] / 3
print(df)
top_students = df.sort_values(
    by="Average",
    ascending=False
)
print(top_students)
```
