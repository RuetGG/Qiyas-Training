```python
import pandas as pd
```

```python
data = {
    "Name": ["Abel", "Sara", "Hana"],
    "Age": [22, 21, 20]
}
df = pd.DataFrame(data)
print(df)
```

```python
dept = ["CS", "IS", "SE", "DS"]
df = pd.DataFrame(dept)
print(df)
```

```python
students = {
"Name": ["Abel", "Sara", "John", "Hana"],
"Age": [22, 21, 23, 20],
"GPA": [3.8, 3.5, 3.2, 3.9],
"Department": ["CS", "IT", "SE", "DS"]

}

df = pd.DataFrame(students)
print(df.shape)
print(df.columns)
print(df.dtypes)
```

```python
# print(df[:3])
print(df.head(3))
print(df.tail(2))
```

```python
print(df["Name"])
print(df[["Name", "GPA"]])
```

```python
print(df.loc[0])
print(df.iloc[2])
```

```python
print(df.loc[1, "GPA"])
```

```python
print(df.iloc[3,1])
```

```python
print(df[df["Age"] > 21])
print(df[df["GPA"] > 3.5])

print(df[(df["Age"] > 20) & (df["GPA"] > 3.5)])
```

```python

print(df[(df["Department"] == "CS") | (df["Department"] == "IT")])
```

```python
df["Percentage"] = df["GPA"] * 25
print(df)
```

```python
df["Pass"] = df["GPA"].apply(
    lambda x: "Pass" if x >= 2.0 else "Fail"
)
print(df)
```

```python
print(df["Age"].mean())
print(df["GPA"].max())
print(df["GPA"].min())
print(df["Age"].sum())
print(df["Name"].count())
```

```python
print(df.sort_values("GPA"))
print(df.sort_values("GPA", ascending=False))
```
