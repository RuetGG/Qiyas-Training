```python
import numpy as np
import pandas as pd
```

```python
data = {
    "Name": ["Abel", "Sara", "John"],
    "Age": [22, np.nan, 20]
}
df = pd.DataFrame(data)
print(df)
```

```python
print(df.isnull())
print(df.isnull().sum())
```

```python
# print(df.fillna(0))
df["Age"] = df["Age"].fillna(df["Age"].mean())
print(df)
```

```python
print(df.dropna())
```

```python
for index, row in df.iterrows():
    print(row["Name"])
    
for row in df.itertuples():
    print(row.Name)
```

```python
students = {
"Name": ["Abel", "Sara", "John", "Hana"],
"Age": [22, 21, 23, 20],
"GPA": [3.8, 3.5, 3.2, 3.9],
"Department": ["CS", "IT", "SE", "DS"]

}

df = pd.DataFrame(students)
df["Scholarship"] = df["GPA"].apply(
    lambda x:
        "Full" if x >= 3.7
        else "Partial" if x >= 3.3 
        else "None")

print(df)
```

```python
def classify(gpa):
    if gpa >= 3.7:
        return "Distinction"
    elif gpa >= 3.0:
        return "Very Good"
    elif gpa >= 2.0:
        return "Good"
    else:
        return "Probation"
    
df["Grade Classification"] = df["GPA"].apply(
    classify
)
print(df)
```
