```python
import pandas as pd

customers = pd.DataFrame({
"Customer_ID":["C001","C002","C003","C005","C006"],
"Customer_Name":["Abel","Sara","John","Helen","Samuel"],
"City":["Addis Ababa","Adama","Jimma","Bahir Dar","Hawassa"]
})

orders = pd.DataFrame({
"Order_ID":[101,102,103,104,105,106],
"Customer_ID":["C001","C002","C003","C001","C004","C006"],
"Product":["Laptop","Phone","Keyboard","Mouse","Monitor","Tablet"],
"Amount":[1200,800,150,45,400,700]
})

employees = pd.DataFrame({
"Employee_ID":[1,2,3,4,5,6],
"Department":["IT","IT","Sales","Sales","HR","HR"],
"Employee":["Abel","Sara","John","Helen","Samuel","David"],
"Salary":[5000,6500,4500,4700,5200,4900]
})

print(customers)
print(orders)
print(employees)
```

```python
inner_join = pd.merge(orders, customers, on='Customer_ID', how='inner')
print(inner_join)
```

```python
left_join = pd.merge(orders, customers, on='Customer_ID', how='left')
left_join[['Customer_Name', 'City']] = (
    left_join[['Customer_Name', 'City']].fillna('Unknown')
)
print(left_join)
```

```python
right_join = pd.merge(orders, customers, on='Customer_ID', how='right')
print(right_join) 
```

```python
outer_join = pd.merge(orders, customers, on='Customer_ID', how='outer')
print(outer_join)
```

```python
print(orders[orders['Amount'] > 500])
```

```python
print(orders[orders['Product'].isin(['Laptop', 'Tablet'])])
```

```python
print(orders[(orders['Amount'] > 200) & ( orders['Amount'] < 900)])
```

```python
print(customers[customers['City'].isin(['Addis Ababa', 'Hawassa'])])
```

```python
print(customers[['Customer_Name', 'City']])
```

```python
print(customers[customers['Customer_Name'].str.startswith('S')])
```

```python
print(orders[orders['Product'] != 'Mouse'])
```

```python
print('Total Sales Amount:', orders['Amount'].sum())
print('Average Sales Amount:', orders['Amount'].mean())
print("Highest Order Amount:", orders['Amount'].max())
print("Lowest Order Amount:", orders['Amount'].min())
print("Total Number of Orders:", len(orders))
```

```python
join = pd.merge(customers, orders, on='Customer_ID', how='outer')

res = (
    join[join['Customer_Name'].isin(['Sara', 'Helen'])]
    ['Amount']
    .fillna(0)
    .mean()
)

print(res)
```
