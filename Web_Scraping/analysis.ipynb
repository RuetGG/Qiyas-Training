```python
import pandas as pd
from collections import Counter
```

```python
df = pd.read_csv("../ethiopian_news.csv")
```

```python
# Description of the dataset
df.head()
```

```python
# Number of titles in the dataset 
df.shape[0]
```

```python
# Longest title 
df["title"].str.len().max()
```

```python
# Shortest title 
df["title"].str.len().min()
```

```python
# Top 10 common words 
Counter(" ".join(df["title"]).lower().split()).most_common(10)
```

```python
# Average title length (character)
df["title"].str.len().mean()
# (word)
length = [len(title.split()) for title in df["title"]]
avg = sum(length)/len(length)
```
